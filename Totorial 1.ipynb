### Step 1

```python
import numpy as np

class Perceptron (object):
    def __init__(self, eta=0.01, n_iter=10):
        # η (eta): the learning rate [cite: 34]
        self.eta = eta
        # n_iter: the number of iterations [cite: 35]
        self.n_iter = n_iter
        self.w_ = None
        self.errors_ = []
```

### Step 2

```python
    def weighted_sum(self, X):
        # Step 2 - Weighted sum [cite: 37]
        # np.dot(X, self.w_[1:]) computes the dot product of input features and weights [cite: 40]
        # self.w_[0] represents the bias term [cite: 40]
        return np.dot(X, self.w_[1:]) + self.w_[0]
```

### Step 3

```python
    def predict(self, X):
        # Step 3 - Step function [cite: 44]
        # returns 1 if weighted sum >= 0.0, else -1 [cite: 46, 48, 49]
        return np.where(self.weighted_sum(X) >= 0.0, 1, -1)
```

### Step 4

```python
    def fit(self, X, y):
        # Step 4 - Create the training method [cite: 50]
        # Initializing the weights to 0 [cite: 53, 75]
        self.w_ = np.zeros(1 + X.shape[1]) 
        self.errors_ = [] 
        
        print("Weights:", self.w_) 

        # training the model for n_iter number of iterations [cite: 57]
        for _ in range(self.n_iter):
            error = 0
            # Loop through each input and corresponding target value [cite: 59]
            for xi, target in zip(X, y):
                # 1. Calculate the predicted value (ŷ) [cite: 60, 77]
                y_pred = self.predict(xi)
                
                # 2. Calculate the weight update [cite: 62]
                # update = η * (y - ŷ) [cite: 63, 64]
                update = self.eta * (target - y_pred)
                
                # 3. Update the weights for each feature [cite: 65, 78]
                self.w_[1:] += update * xi
                
                # Update the bias (w0) [cite: 70, 78]
                self.w_[0] += update
                
                # Increment error count if prediction was incorrect [cite: 73]
                error += int(update != 0.0)
            
            # Store the error count after each iteration [cite: 74]
            self.errors_.append(error)
        return self
```

### Step 5 Loading and preprocessing

```python
import pandas as pd
from sklearn.utils import shuffle
from sklearn.model_selection import train_test_split

# Step 5: Loading Data 
# Reading the Iris dataset from the given URL [cite: 121, 126, 128]
df = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data', header=None)

# Shuffling the rows to randomize the order [cite: 129]
df = shuffle(df)
print(df.head())

# Extracting the first four columns (features) [cite: 134, 135]
X = df.iloc[:, 0:4].values

# Extracting the fifth column (target) [cite: 136, 137]
y = df.iloc[:, 4].values

# Splitting the data into train (75%) and test (25%) set [cite: 143, 144, 146]
train_data, test_data, train_labels, test_labels = train_test_split(X, y, test_size=0.25)

# Encoding the labels: 1 for setosa, -1 otherwise [cite: 147, 149]
train_labels = np.where(train_labels == 'Iris-setosa', 1, -1)
test_labels = np.where(test_labels == 'Iris-setosa', 1, -1)
```

### Step 6. Training and Performance Measurement (Scikit-Learn)

```python
from sklearn.linear_model import Perceptron as SkPerceptron
from sklearn.metrics import accuracy_score

# Step 6: Training the perceptron using scikit-learn [cite: 160, 161]
# Initializing with eta0=0.1 and max_iter=10 [cite: 163, 165]
perceptron = SkPerceptron(eta0=0.1, max_iter=10)
perceptron.fit(train_data, train_labels)
```

```python
# Step 7: Making predictions on test data [cite: 168, 169]
test_preds = perceptron.predict(test_data)
```

```python
# Step 8: Measuring Performances [cite: 176, 184]
accuracy = accuracy_score(test_preds, test_labels)
# Print accuracy as a percentage [cite: 187, 188]
print("Accuracy:", round(accuracy, 2) * 100)
```

```python
import numpy as np

class PerceptronSigmoid(object):
    def __init__(self, eta=0.01, n_iter=10):
        self.eta = eta
        self.n_iter = n_iter
        self.w_ = None
        self.errors_ = []

    def sigmoid(self, z):
        # Task 2: Sigmoid Activation Function
        return 1.0 / (1.0 + np.exp(-z))

    def weighted_sum(self, X):
        return np.dot(X, self.w_[1:]) + self.w_[0]

    def predict(self, X):
        # Using Sigmoid output with 0.5 as the decision threshold
        # Maps to 1 and -1 to stay consistent with existing data labels
        phi_z = self.sigmoid(self.weighted_sum(X))
        return np.where(phi_z >= 0.5, 1, -1)

    def fit(self, X, y):
        self.w_ = np.zeros(1 + X.shape[1])
        self.errors_ = []
        
        for _ in range(self.n_iter):
            error_count = 0
            for xi, target in zip(X, y):
                y_pred = self.predict(xi)
                update = self.eta * (target - y_pred)
                self.w_[1:] += update * xi
                self.w_[0] += update
                error_count += int(update != 0.0)
            self.errors_.append(error_count)
        return self

# Assuming train_data and train_labels are defined from previous steps
# Training the Sigmoid version
model_sig = PerceptronSigmoid(n_iter=10)
model_sig.fit(train_data, train_labels)

print("--- Task 2 Output ---")
print("Weights after Sigmoid training:", model_sig.w_)
print("Errors per iteration:", model_sig.errors_)
```

```python
import numpy as np
import pandas as pd
from sklearn.utils import shuffle
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# --- UPDATED PERCEPTRON CLASS ---
class PerceptronSigmoid01(object):
    def __init__(self, eta=0.01, n_iter=10):
        self.eta = eta
        self.n_iter = n_iter
        self.w_ = None
        self.errors_ = []

    def sigmoid(self, z):
        """Task 2: Sigmoid Activation Function"""
        return 1.0 / (1.0 + np.exp(-z))

    def weighted_sum(self, X):
        return np.dot(X, self.w_[1:]) + self.w_[0]

    def predict(self, X):
        """Task 3: Predict using 0 and 1 labels"""
        phi_z = self.sigmoid(self.weighted_sum(X))
        return np.where(phi_z >= 0.5, 1, 0)

    def fit(self, X, y):
        self.w_ = np.zeros(1 + X.shape[1])
        self.errors_ = []
        
        for _ in range(self.n_iter):
            error_count = 0
            for xi, target in zip(X, y):
                y_pred = self.predict(xi)
                update = self.eta * (target - y_pred)
                self.w_[1:] += update * xi
                self.w_[0] += update
                error_count += int(update != 0.0)
            self.errors_.append(error_count)
        return self

# --- THE COMPLETE PIPELINE ---

# 1. Load Data
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
df = pd.read_csv(url, header=None)
df = shuffle(df)

# 2. Prepare Features and Labels
X = df.iloc[:, 0:4].values
y = df.iloc[:, 4].values

# 3. Train/Test Split
train_data, test_data, train_labels, test_labels = train_test_split(X, y, test_size=0.25)

# 4. Task 3: Encode labels as 1 for 'Iris-setosa' and 0 for others
train_labels_01 = np.where(train_labels == 'Iris-setosa', 1, 0)
test_labels_01 = np.where(test_labels == 'Iris-setosa', 1, 0)

# 5. Initialize and Train Model
model = PerceptronSigmoid01(eta=0.1, n_iter=20)
model.fit(train_data, train_labels_01)

# 6. Final Evaluation
test_preds = model.predict(test_data)
accuracy = accuracy_score(test_labels_01, test_preds)

print("--- Task 3: Full Pipeline Results ---")
print(f"Accuracy: {round(accuracy, 2) * 100}%")
print(f"Final Weights: {model.w_}")
print(f"Error count per iteration: {model.errors_}")

# Task 1: Manual Verification
print("\n--- Manual Check (First 5 Test Samples) ---")
print(f"Predicted: {test_preds[:5]}")
print(f"Actual:    {test_labels_01[:5]}")
```
